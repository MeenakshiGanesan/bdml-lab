```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import matplotlib.pyplot as plt

(x_train, y_train), (x_test, y_test) = datasets.cifar10.load_data()
x_train, x_test = x_train / 255.0, x_test / 255.0

model_ff = models.Sequential([
    layers.Flatten(input_shape=(32, 32, 3)),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')
])
model_ff.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
ff_history = model_ff.fit(x_train, y_train, epochs=5, validation_data=(x_test, y_test))

model_cnn = models.Sequential([
    layers.Conv2D(32, (3,3), activation='relu', input_shape=(32, 32, 3)),
    layers.MaxPooling2D((2,2)),
    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D((2,2)),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])
model_cnn.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
cnn_history = model_cnn.fit(x_train, y_train, epochs=5, validation_data=(x_test, y_test))

plt.plot(cnn_history.history['accuracy'], label='Training Accuracy')
plt.plot(cnn_history.history['val_accuracy'], label='Validation Accuracy')
plt.title('CNN Accuracy')
plt.legend()
plt.show()
```

```python
# ------------------------------
# Feedforward NN vs CNN on MNIST
# ------------------------------

import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.datasets import mnist
import matplotlib.pyplot as plt

# 1️⃣ Load Dataset
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Normalize pixel values
X_train, X_test = X_train / 255.0, X_test / 255.0

# ------------------------------
# Feedforward Neural Network (FNN)
# ------------------------------

# Flatten the 28x28 images into 784 features
X_train_flat = X_train.reshape(-1, 784)
X_test_flat = X_test.reshape(-1, 784)

fnn_model = models.Sequential([
    layers.Dense(128, activation='relu', input_shape=(784,)),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

fnn_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
fnn_history = fnn_model.fit(X_train_flat, y_train, epochs=5, validation_data=(X_test_flat, y_test), verbose=1)

# ------------------------------
# Convolutional Neural Network (CNN)
# ------------------------------

# Reshape for CNN: add channel dimension
X_train_cnn = X_train.reshape(-1, 28, 28, 1)
X_test_cnn = X_test.reshape(-1, 28, 28, 1)

cnn_model = models.Sequential([
    layers.Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)),
    layers.MaxPooling2D(2,2),
    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D(2,2),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')
])

cnn_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
cnn_history = cnn_model.fit(X_train_cnn, y_train, epochs=5, validation_data=(X_test_cnn, y_test), verbose=1)

# ------------------------------
# Compare Performances
# ------------------------------

plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.plot(fnn_history.history['accuracy'], label='FNN Train Acc')
plt.plot(fnn_history.history['val_accuracy'], label='FNN Val Acc')
plt.plot(cnn_history.history['accuracy'], label='CNN Train Acc')
plt.plot(cnn_history.history['val_accuracy'], label='CNN Val Acc')
plt.title('Model Accuracy Comparison')
plt.xlabel('Epochs'); plt.ylabel('Accuracy'); plt.legend()

plt.subplot(1,2,2)
plt.plot(fnn_history.history['loss'], label='FNN Train Loss')
plt.plot(fnn_history.history['val_loss'], label='FNN Val Loss')
plt.plot(cnn_history.history['loss'], label='CNN Train Loss')
plt.plot(cnn_history.history['val_loss'], label='CNN Val Loss')
plt.title('Model Loss Comparison')
plt.xlabel('Epochs'); plt.ylabel('Loss'); plt.legend()

plt.show()

# ------------------------------
# Evaluate and Print Final Results
# ------------------------------
fnn_test_acc = fnn_model.evaluate(X_test_flat, y_test, verbose=0)[1]
cnn_test_acc = cnn_model.evaluate(X_test_cnn, y_test, verbose=0)[1]

print(f"Feedforward NN Test Accuracy: {fnn_test_acc:.4f}")
print(f"Convolutional NN Test Accuracy: {cnn_test_acc:.4f}")
```
